```python
import matplotlib.pyplot as plt
from PIL import Image
import os

# ==========================
# Global font configuration
# ==========================
plt.rcParams.update({
    "font.family": "Arial",
    "font.size": 12,
})

# ==========================
# Working directory and paths
# ==========================
base_dir = r"C:\Users\janat\Desktop\NSQIP_New\plots"
images = [
    os.path.join(base_dir, "pca_before_smote_trainingdata.png"),
    os.path.join(base_dir, "SMOTE.png"),
    os.path.join(base_dir, "feature_importance.png"),
    os.path.join(base_dir, "SHAP.png")
]
output_path = os.path.join(base_dir, "Figure1.png")

# ==========================
# Panel labels and captions
# ==========================
labels = ['A', 'B', 'C', 'D']
captions = [
    "No SMOTE PCA",
    "SMOTE PCA",
    "Feature Importance",
    "SHAP Analysis"
]

# ==========================
# Resize images to same height while preserving aspect ratio
# ==========================
target_height = 800  # pixels
resized_images = []

for img_path in images:
    img = Image.open(img_path)
    aspect_ratio = img.width / img.height
    new_width = int(target_height * aspect_ratio)
    img_resized = img.resize((new_width, target_height))
    resized_images.append(img_resized)

# ==========================
# Create a 2x2 subplot layout
# ==========================
fig, axes = plt.subplots(2, 2, figsize=(8, 8.5), dpi=300)
axes = axes.flatten()

for i, ax in enumerate(axes):
    if i < len(resized_images):
        ax.imshow(resized_images[i])
        ax.axis("off")
        ax.set_aspect('auto')

        # Add panel label
        ax.text(
            -0.05, 1.05, labels[i],
            transform=ax.transAxes,
            fontsize=16,
            va='top',
            ha='left',
            bbox=dict(facecolor="white", alpha=0.6, edgecolor="none", pad=2)
        )

        # Add caption
        ax.set_title(captions[i], fontsize=12, pad=8)
    else:
        ax.axis("off")

# ==========================
# Final layout and export
# ==========================
plt.tight_layout(h_pad=1.2, w_pad=1.2)
plt.savefig(output_path, dpi=300, bbox_inches="tight")
plt.show()

print(f"✅ Figure saved at: {output_path}")
```

```python
import matplotlib.pyplot as plt
from PIL import Image
import os

# ==========================
# Global font configuration
# ==========================
plt.rcParams.update({
    "font.family": "Arial",
    "font.size": 12,
})

# ==========================
# Base directory and image paths
# ==========================
base_dir = r"C:\Users\janat\Desktop\NSQIP_New\plots"
images = [
    os.path.join(base_dir, "ROC_smote.png"),
    os.path.join(base_dir, "ROC_no_SMOTE.png"),
    os.path.join(base_dir, "Blind_set2_SMOTE.png"),
    os.path.join(base_dir, "AUC_comparison.png"),
     os.path.join(base_dir, "XGb_AUC.png"),
    os.path.join(base_dir, "XGb_boxplot.png")
   
]

# ==========================
# Labels and captions
# ==========================
labels = ['A', 'B', 'C', 'D', 'E', 'F']
captions = [
    "ROC: SMOTE",
    "ROC: No SMOTE",
    "ROC: SMOTE Models Additional Blind Set",
    "All Prediction Models AUC Comparison",
    "XGB: AUC Comparison",
    "XGB: Probability Value Distribution"
]

output_path = os.path.join(base_dir, "Figure2.png")

# ==========================
# Load images without resizing
# ==========================
loaded_images = []
for img_path in images:
    try:
        img = Image.open(img_path)
        loaded_images.append(img)
    except Exception as e:
        print(f"⚠️ Could not load image {img_path}: {e}")

# ==========================
# Create 3 rows × 2 columns subplot layout
# ==========================
fig, axes = plt.subplots(3, 2, figsize=(14, 18), dpi=300)
axes = axes.flatten()

for i, ax in enumerate(axes):
    if i < len(loaded_images):
        ax.imshow(loaded_images[i])
        ax.axis("off")

        # Panel label (e.g., A, B, C, ...)
        ax.text(
            -0.05, 1.05, labels[i],
            transform=ax.transAxes,
            fontsize=12,
            va='top',
            ha='left',
            bbox=dict(facecolor="white", alpha=0.5, edgecolor="none", pad=2)
        )

        # Caption (e.g., ROC: SMOTE, etc.)
        ax.set_title(captions[i], fontsize=12, pad=6)
    else:
        ax.axis("off")  # Leave empty if not enough images

# ==========================
# Final layout and export
# ==========================
plt.tight_layout(h_pad=2.0, w_pad=1.5)
plt.savefig(output_path, dpi=330, bbox_inches="tight")
plt.show()

print(f"✅ Figure saved at: {output_path}")
```
